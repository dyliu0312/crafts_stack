```python
from crafts_stack.hpmap import hp, np, build_healpix_map, cut_hpmap, get_map_center_range
from crafts_stack.plot import plot_hp_grid
```

```python
nside = 32
npix = hp.nside2npix(nside)
values = np.random.rand(npix)
pixels = range(npix)
```

```python
hpmap = build_healpix_map(nside, pixels, values)
type(hpmap)
```

```python
kw_graticule = dict(color='r', linestyle='dashed', linewidth=0.5, dpar=30, dmer=30)
plot_hp_grid(hpmap, graticule_kw=kw_graticule)
```

```python
pix, val = cut_hpmap(nside, pixels, values, [150, 255], [40, 45])
cut_hp = build_healpix_map(nside, pix, val)
```

```python
plot_hp_grid(cut_hp)
```

```python
rot, lonra, latra = get_map_center_range(nside, pix)
```

```python
# hp_kw = dict(lonra=lonra, latra=latra)
hp_kw = dict(rot=rot, lonra=lonra, latra=latra)
plot_hp_grid(cut_hp, hpfunc=hp.cartview, hpfunc_kw=hp_kw)
```
